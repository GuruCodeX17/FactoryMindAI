Import the libraries

```python
import fitz
import json
from pathlib import Path

print("PyMuPDF version:", fitz.version)
print("Libraries imported successfully!")
```

Load your PDF

```python
PROJECT_ROOT = Path.cwd().parent

PDF_FOLDER = (
    PROJECT_ROOT
    / "knowledge_base"
    / "raw"
    / "maintenance_guides"
)

pdf_files = list(PDF_FOLDER.glob("*.pdf"))

print("PDF files found:", len(pdf_files))

for file in pdf_files:
    print(file.name)
```

Extract text page by page

```python
if not pdf_files:
    raise FileNotFoundError("No PDF found in maintenance_guides!")

pdf_path = pdf_files[0]

document = fitz.open(pdf_path)

extracted_pages = []

for page_number, page in enumerate(document, start=1):

    text = page.get_text("text")

    extracted_pages.append({
        "source": pdf_path.name,
        "page": page_number,
        "text": text
    })

document.close()

print("Total pages:", len(extracted_pages))
print("PDF extraction completed!")
```

Inspect extracted content

```python
for page in extracted_pages[:3]:

    print("\nPage:", page["page"])
    print("-" * 60)

    print(page["text"][:1000])
```

Save the extracted text

```python
OUTPUT_FOLDER = (
    PROJECT_ROOT
    / "knowledge_base"
    / "processed"
)

OUTPUT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

output_path = OUTPUT_FOLDER / "extracted_documents.jsonl"

with open(output_path, "w", encoding="utf-8") as file:

    for page in extracted_pages:
        file.write(json.dumps(page, ensure_ascii=False) + "\n")

print("Saved successfully:", output_path)
```
